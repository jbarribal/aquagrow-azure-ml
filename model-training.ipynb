```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Lasso
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_squared_error
```

```python
# Load the data into a pandas DataFrame
df = pd.read_csv('append_data.csv')

# Split the data into input (X) and output (y) variables
X = df.iloc[:, :-1]
y = df.iloc[:, -1]

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
```

```python
# Create a LinearRegression model and fit it to the training data
lin_model = LinearRegression()
lin_model.fit(X_train, y_train)

# Predict the output variable using the model on the test data
lin_y_pred = lin_model.predict(X_test)

# Calculate the R^2 score and print it
lin_r2 = r2_score(y_test, lin_y_pred)
print('R^2 score:', lin_r2)

# Calculate the RMSE and print it
lin_mse = mean_squared_error(y_test, lin_y_pred)
lin_rmse = lin_mse ** 0.5
print('RMSE:', lin_rmse)
```

```python
# Create an SVR model and fit it to the training data
svr_model = SVR(kernel='linear')
svr_model.fit(X_train, y_train)

# Predict the output variable using the model on the test data
svr_y_pred = svr_model.predict(X_test)

# Calculate the R^2 score and print it
svr_r2 = r2_score(y_test, svr_y_pred)
print('R^2 score:', svr_r2)

# Calculate the RMSE and print it
svr_mse = mean_squared_error(y_test, svr_y_pred)
svr_rmse = svr_mse ** 0.5
print('RMSE:', svr_rmse)
```

```python
# Create a Lasso model and fit it to the training data
las_model = Lasso(alpha=0.1)
las_model.fit(X_train, y_train)

# Predict the output variable using the model on the test data
las_y_pred = las_model.predict(X_test)

# Calculate the R^2 score and print it
las_r2 = r2_score(y_test, las_y_pred)
print('R^2 score:', las_r2)

# Calculate the RMSE and print it
las_mse = mean_squared_error(y_test, las_y_pred)
las_rmse = las_mse ** 0.5
print('RMSE:', las_rmse)
```

```python
# Create a Random Forest model and fit it to the training data
rfr_model = RandomForestRegressor(n_estimators=100, random_state=0)
rfr_model.fit(X_train, y_train)

# Predict the output variable using the model on the test data
rfr_y_pred = rfr_model.predict(X_test)

# Calculate the R^2 score and print it
rfr_r2 = r2_score(y_test, rfr_y_pred)
print('R^2 score:', rfr_r2)

# Calculate the RMSE and print it
rfr_mse = mean_squared_error(y_test, rfr_y_pred)
rfr_rmse = rfr_mse ** 0.5
print('RMSE:', rfr_rmse)
```
